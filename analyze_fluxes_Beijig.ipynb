```python
data_path = r"D:\!Work\IFA\Beijing\tower data\2024.6\Xianghe tower\32m"
files = glob.glob(os.path.join(data_path, "*.dat"))
df_rur = pd.DataFrame()
for file in tqdm(files):
    cur_df = pd.read_csv(file, header=1, skiprows=range(2,4), parse_dates=['TIMESTAMP'], index_col = 'TIMESTAMP')
    df_rur = pd.concat((df_rur, cur_df))
    
```

```python
df_rur['Cp'] = pd.to_numeric(df_rur['Cp'], errors='coerce')
df_rur['rho_a_Avg'] = pd.to_numeric(df_rur['rho_a_Avg'], errors='coerce')
df_rur['rho_d_Avg'] = pd.to_numeric(df_rur['rho_d_Avg'], errors='coerce')
```
