```python
import os, sys
import glob
import shutil
```

```python
data_dir = r"D:\!Work\IFA\!data\Urban_fluxes_NE\processed"
repo_dir = r"D:\!Work\!git\datasets\UrbanFluxesNE"

cities2copy = ['Tomsk', 'Moscow', 'Beijing']
corr_str = {'Tomsk':'new_py'}

for city in cities2copy:
    if os.path.isdir(repo_dir + '/' + city): 
        shutil.rmtree (repo_dir + '/' + city)
    os.makedirs(repo_dir + '/' + city, exist_ok=True)
    files = glob.glob(data_dir + '/' + city + (('_' + corr_str[city]) if city in corr_str else '') + '/L1/*.csv')
    display(files)

    for file in files:
        fname = os.path.basename(file)
        if 'ERA5' in fname or 'LOD' in fname: continue
        shutil.copy (file, repo_dir + '/' + city + '/' + fname)

shutil.copy (data_dir + '\sitedata_basic.csv', repo_dir + '\site_data_basic.csv')
shutil.copy (data_dir + '\sitedata.csv', repo_dir + '\site_data.csv')
```
